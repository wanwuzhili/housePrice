```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
import joblib
```

```python
train_data = pd.read_csv("../data/raw/train.csv")
```

```python
train_features = pd.concat((train_data.iloc[:, 1], train_data.iloc[:, 3:]), axis=1)
```

```python
def clean_feature(feature, num_rare):
    feature = feature.fillna("unknown")
    feature = feature.astype(str).str.lower().str.strip()
    counts = feature.value_counts()
    rare_type = counts[counts < num_rare].index
    return feature.replace(rare_type, "other")
```

```python
type_cleaned = clean_feature(train_features.Type, 5)
```

```python
type_cleaned.nunique()
```

```python
def encoder(feature):
    le = LabelEncoder()
    feature = le.fit_transform(feature)
    return feature, le
```

```python
type_encode, type_le = encoder(type_cleaned)
```
